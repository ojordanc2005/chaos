```python
# to animate anything, you need the basic structure of a plotting function to be called and
# an animate function.

# my ax
fig = plt.figure()
ax = plt.axes()
ax.set_facecolor('black')

num_walks = 60
ax.set_xlim(-num_walks/2 + 1, num_walks/2 + 1)
ax.set_ylim(-num_walks/2 + 1, num_walks/2 + 1)

x_vals = np.linspace(-num_walks/2 + 0.5, num_walks/2 + 0.5, num_walks)
y_vals = np.linspace(-num_walks/2 + 0.5, num_walks/2 + 0.5, num_walks)

ax.set_xticks((x_vals), labels='')
ax.set_yticks((y_vals), labels='')

ax.grid()

# starting cube is at (0,0)
# evolution is based on number generation: 1, 2, 3, 4 -> N, E, S, W
# i'll use a grid structure to fill in squares, starting from the ordered (x,y) and extending positively

#main loop
x = 0
y = 0

x_array = []
y_array = []

for i in range(num_walks):
    move = np.random.randint(1,5)
    #print(move)

    x_fill = [x, x, x+1, x+1]
    y_fill = [y, y+1, y+1, y]

    ax.fill(x_fill, y_fill, color='white')

    if (move == 1):
        # move North
        y = y + 1
    if (move == 2):
        # move East
        x = x + 1
    if (move == 3):
        # move South
        y = y - 1
    if (move == 4):
        # move West
        x = x - 1

    x_array.append(x)
    y_array.append(y)

plt.show()    


x_array = np.array(x_array)
y_array = np.array(y_array)
```

```python
# my ax
fig = plt.figure()
ax = plt.axes()
ax.set_facecolor('black')

ax.set_xlim(-num_walks/2 + 1, num_walks/2 + 1)
ax.set_ylim(-num_walks/2 + 1, num_walks/2 + 1)

x_vals = np.linspace(-num_walks/2 + 0.5, num_walks/2 + 0.5, num_walks)
y_vals = np.linspace(-num_walks/2 + 0.5, num_walks/2 + 0.5, num_walks)

ax.set_xticks((x_vals), labels='')
ax.set_yticks((y_vals), labels='')

ax.grid()

# example of animating ------------------------------------------------
def plotSomeStuff(x_array, y_array, num_walks):

    def animate(walk_num):
        x = x_array[walk_num]
        y = y_array[walk_num]

        x_fill = [x, x, x+1, x+1]
        y_fill = [y, y+1, y+1, y]

        ax.fill(x_fill, y_fill, color='white')

        return
    
    anim = FuncAnimation(fig, animate, frames=num_walks, interval=70)
    display(HTML(anim.to_jshtml()))
    plt.close()

plotSomeStuff(x_array, y_array, num_walks)
```
